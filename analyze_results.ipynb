# Analysis of Inference Results

This notebook analyzes the CSV results generated by `run_inference.py`.
It calculates metrics (Recall, Accuracy, F1) and displays confusion matrices.

```python
import pandas as pd
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_recall_fscore_support
import os
```

```python
# Configuration
RESULT_DIR = 'result'
RUN_DATE = '0426'
RUN_INDEX = 173
MODEL_NAME = 'secbert'

# Construct filenames based on inference script naming convention
file_prefix = f'result/{MODEL_NAME}_{RUN_DATE}_{RUN_INDEX}_result'

print(f'Reading results from prefix: {file_prefix}')
```

```python
# Load Results
try:
    time_result = pd.read_csv(f'{file_prefix}_time.csv')
    tag_result = pd.read_csv(f'{file_prefix}_tag.csv')
    scale_result = pd.read_csv(f'{file_prefix}_scale.csv')
    negative_result = pd.read_csv(f'{file_prefix}_negative.csv')
    
    # Optional: Load Fact result if available
    fact_path = f'{file_prefix}_fact.csv'
    if os.path.exists(fact_path):
        fact_result = pd.read_csv(fact_path)
    else:
        fact_result = None
        
    print('✅ Loaded all result CSVs.')
except FileNotFoundError as e:
    print(f'❌ Error loading files: {e}')
    print('Please ensure run_inference.py has finished successfully.')
```

```python
def print_metrics(df, attr_name, true_col=None, pred_col=None):
    if df is None:
        print(f'No data for {attr_name}')
        return
        
    print(f'\{attr_name.upper()} Classification Metrics:')
    if true_col is None: true_col = f'true_{attr_name}'
    if pred_col is None: pred_col = f'pred_{attr_name}'
    
    y_true = df[true_col]
    y_pred = df[pred_col]
    
    acc = accuracy_score(y_true, y_pred)
    precision_w, recall_w, f1_w, _ = precision_recall_fscore_support(y_true, y_pred, average='weighted', zero_division=0)
    precision_m, recall_m, f1_m, _ = precision_recall_fscore_support(y_true, y_pred, average='macro', zero_division=0)
    
    print(f'Accuracy: {acc:.4f}')
    print(f'Precision (weighted): {precision_w:.4f}, (macro): {precision_m:.4f}')
    print(f'Recall (weighted): {recall_w:.4f}, (macro): {recall_m:.4f}')
    print(f'F1 Score (weighted): {f1_w:.4f}, (macro): {f1_m:.4f}')
    
    print('Confusion Matrix:')
    print(confusion_matrix(y_true, y_pred))
    
    print('Classification Report:')
    print(classification_report(y_true, y_pred, zero_division=0))
```

```python
# Analyze TIME
print_metrics(time_result, 'time')
```

```python
# Analyze SCALE
print_metrics(scale_result, 'scale')
```

```python
# Analyze NEGATIVE
print_metrics(negative_result, 'negative')
```

```python
# Analyze TAG
print_metrics(tag_result, 'tag')
```

```python
# Analyze FACT (Regression)
if fact_result is not None:
    print('\FACT Regression Metrics:')
    from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
    
    y_true = fact_result['true_fact']
    y_pred = fact_result['pred_fact']
    
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    
    print(f'MSE: {mse:.4f}')
    print(f'RMSE: {rmse:.4f}')
    print(f'MAE: {mae:.4f}')
    print(f'R² Score: {r2:.4f}')
```

## Save Analysis Output
Per requirement: 'let data analysis output to csv'
We will save a summary of metrics to a CSV file.

```python
metrics_data = []

def get_metrics_dict(df, attr_name):
    if df is None: return None
    y_true = df[f'true_{attr_name}']
    y_pred = df[f'pred_{attr_name}']
    acc = accuracy_score(y_true, y_pred)
    p_w, r_w, f1_w, _ = precision_recall_fscore_support(y_true, y_pred, average='weighted', zero_division=0)
    p_m, r_m, f1_m, _ = precision_recall_fscore_support(y_true, y_pred, average='macro', zero_division=0)
    return {'Task': attr_name, 'Accuracy': acc, 'Precision_Weighted': p_w, 'Recall_Weighted': r_w, 'F1_Weighted': f1_w, 'Precision_Macro': p_m, 'Recall_Macro': r_m, 'F1_Macro': f1_m}

for df, name in [(time_result, 'time'), (scale_result, 'scale'), (negative_result, 'negative'), (tag_result, 'tag')]:
    m = get_metrics_dict(df, name)
    if m: metrics_data.append(m)

metrics_df = pd.DataFrame(metrics_data)
output_csv = f'analysis_metrics_{RUN_DATE}_{RUN_INDEX}.csv' if 'RUN_DATE' in locals() else 'analysis_metrics.csv'
metrics_df.to_csv(output_csv, index=False)
print(f'Saved analysis metrics to {output_csv}')
metrics_df
```
